# Week 1: Student Performance Analysis
This notebook analyzes students' scores in Math, Reading, and Writing.  
We explore how gender, lunch type, parental education, and test preparation course affect performance.


# Step 1: Load the Dataset

```python
import pandas as pd
df = pd.read_csv("StudentsPerformance.csv")
df.head()
```

```python
df.head()
```

# Step 2: Explore the Data

```python
df.info()
df.describe()

for col in ['gender', 'parental level of education', 'lunch', 'test preparation course']:
    print(f"\n{col} distribution:")
    print(df[col].value_counts())
```

# Step 3: Check for Missing Values

```python
df.isnull().sum()
```

# Step 4: Analyze Score Patterns
We analyze average scores grouped by categorical features.


# .Average scores by gender
# 'Average scores by lunch type
# .Average scores by test preparation course
# .Average scores by parental level of education

```python

df.groupby("gender")[["math score","reading score","writing score"]].mean()
df.groupby("lunch")[["math score","reading score","writing score"]].mean()
df.groupby("test preparation course")[["math score","reading score","writing score"]].mean()
df.groupby("parental level of education")[["math score","reading score","writing score"]].mean()
```

# Step 5: Visualize the Data
We use barplots, boxplots, histograms, and heatmaps to understand trends.


```python
import seaborn as sns
import matplotlib.pyplot as plt

# Gender vs Scores (Boxplot)
sns.boxplot(x="gender", y="math score", data=df)
plt.show()

# Lunch vs Reading Score (Barplot)
sns.barplot(x="lunch", y="reading score", data=df)
plt.show()

# Test Preparation vs Writing Score
sns.barplot(x="test preparation course", y="writing score", data=df)
plt.show()

# Distribution of Math Scores
sns.histplot(df["math score"], kde=True, bins=20)
plt.show()
```

```python
# 4. Correlation Heatmap
plt.figure(figsize=(6,5))
sns.heatmap(df[['math score','reading score','writing score']].corr(), annot=True, cmap='coolwarm')
plt.title('Correlation Between Scores')
plt.show()
```

#Findings

Test Preparation: Students who completed the test preparation course scored higher in reading and writing, indicating that preparation positively impacts performance.

Lunch Type: Students with a standard lunch generally performed better than those with free/reduced lunch, suggesting a slight correlation between nutrition and academic performance.

Gender Differences: Males scored slightly higher in math, while females had slightly higher reading and writing scores.

Parental Education: Higher parental education levels (e.g., Master's degree, Bachelor's degree) correlate with better student performance across all subjects.

Score Correlation: Math, reading, and writing scores are strongly correlated, showing that students who perform well in one subject tend to perform well in others.

#Conclusion

This analysis highlights how student performance varies with factors like gender, lunch type, parental education, and test preparation. Key observations include the positive impact of test preparation and the influence of parental education on scores.

Visualizations such as barplots, boxplots, histograms, and correlation heatmaps helped uncover these trends. This beginner-friendly project demonstrates the importance of exploratory data analysis (EDA) in understanding real-world datasets and communicating insights clearly.
